```python
n = 5
v = [0] * n

for i in range(n):
    v[i] = int(input("Digite um valor: "))

for j in range(n):
    menor = j

    for k in range(j + 1, n):
        if v[menor] > v[k]:
            menor = k

    aux = v[menor]
    v[menor] = v[j]
    v[j] = aux

print(v)
```

```python
import math
mPontos = []
mDistancia = []

for i in range(1,5):
    mPontos.append([0]*2)
    mDistancia.append([0]*i)

print("Digite 4 pontos pertencentes ao plano cartesiano: ")
for i in range(0,4):
    mPontos[i][0] = int(input("x: "))
    mPontos[i][1] = int(input("y: "))
for i in range(0,4):
    for j in range(i,4):
      mDistancia[j][i] = math.sqrt((mPontos[i][0] - mPontos[j][0])**2 + (mPontos[i][1] - mPontos[j][1])**2)
for i in range(0,4):
    for j in range(0,i):
        print(mDistancia[i][j], end=" " )
    print(end="\n")



```

```python
#Funções
def primo(x): #cabeçalho de função
    p = True  #corpo da função
    for i in range(2,x):
        if (x % i) == 0:
            p = False
    return p

x = primo(x=14)
print(x)
```

```python
def fatorial( x ): #calcular fatorial.
    fat_x = 1
    for y in range(1, x+1):
        fat_x = fat_x * y
    return fat_x

def somaSeries(n):
    s = 1
    for a in range(1, n+1):
      s = s + 1/fatorial(a)
    return s

print(fatorial(3))
```

```python
def funcao(v, x):
    for i in range(len(v)):
        if v[i] == x:
            return i

    return -1

posicao = funcao([1, 2, 3, 4, 5, 6, 7, 8], 5)
print(posicao)
```
