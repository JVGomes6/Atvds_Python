1. O custo ao consumidor de um carro novo é a soma do custo de fábrica com a percentagem do distribuidor e dos impostos (aplicados, primeiro os impostos sobre o custo de fábrica, e depois a percentagem do distribuidor sobre o resultado). Supondo que a percentagem do distribuidor seja de 28% e os impostos 45%. Escrever um algoritmo que leia o custo de fábrica de um carro e informe o custo ao consumidor do mesmo.

2. Faça um algoritmo que receba um número e mostre uma mensagem caso
este número seja maior que 10.

3. Escrever um algoritmo que leia dois valores inteiro distintos e informe
qual é o maior.

4. Faça um algoritmo que receba um número e diga se este número está no
intervalo entre 100 e 200.

5. Escrever um algoritmo que leia o nome e as três notas obtidas por um
aluno durante o semestre. Calcular a sua média (aritmética), informar o
nome e sua menção aprovado (media >= 7), // Reprovado (media < 4) e
Recuperação (media entre 4 a 6.9).

```python
#Exercício 1

cdf= float(input("digite o custo do carro: "))

#distribuidor = 28%:
#impostos de fábrica = 45%

#Cálculo

result=cdf+(cdf*0.45)
real=result+(result*0.28)

#Tela

print(f"O Valor do carro ao consumidor final: {real}")
```
