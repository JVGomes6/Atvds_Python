**Enquanto (while)**

```
while (condição):
  < Enquanto a condição for verdadeira
  faça este grupo de comandos >

< Quando a condição for falsa, saia do laço e continue >
```


    

````python
materia = input("Insira o nome da melhor disciplina do curso")
while materia!="programação":
	materia = input("Resposta errada, tente novamente:")
print("Resposta Correta!")
````

**Laço finito (número específico de execuções)**

Uso de variável auxiliar para "contar" o número de execuções.

Pode-se utilizar
- While = enquanto - faça n vezes
- For

````python
N = int(input("Insira o valor de N: "))
cont = 0
while cont<N:
	print("Eu sei programar!")
	cont+=1
````

**Comando Para (for)**

Parecido com o While, mas o próprio comando controla a atualização da variável auxiliar.

```
for (variável) in (range):
  < Para todos os números no range
  ou intervalo estipulado, execute
  os comandos deste bloco >

< Quando finalizar, saia do laço e continue >
```

````python
N = int(input("Insira o valor de N: "))
for cont in range(N):
	print("Eu sei programar!")
````

**Ler 80 números e ao final informar quantos número(s) est(á)ão no intervalo entre 10 (inclusive) e 150 (inclusive).**

````python
contador = 0

for i in range(80):

    while True:
        entrada = input(f"Digite o {i+1}º número: ")

        if entrada.strip() == "":
            print("Erro: campo vazio. Digite um número.")
            continue

        try:
            num = int(entrada)
            break
        except ValueError:
            print("Erro: digite apenas números inteiros.")

    if 10 <= num <= 150:
        contador += 1

print(f"Existem {contador} números no intervalo entre 10 e 150.")
````

````python
maiores = 0
menores = 0

for i in range(75):
    idade = int(input(f"Digite a idade da {i+1}ª pessoa: "))

    if idade >= 18:
        print("Maior de idade")
        maiores += 1
    else:
        print("Menor de idade")
        menores += 1

print(f"Total de maiores: {maiores}")
print(f"Total de menores: {menores}")
````

````python
homens = 0
mulheres = 0

for i in range(56):

    nome = input("Digite o nome: ")
    sexo = input("Digite o sexo (M/F): ").upper()

    if sexo == "M":
        print(nome, "- Homem")
        homens += 1

    elif sexo == "F":
        print(nome, "- Mulher")
        mulheres += 1

    else:
        print("Sexo inválido")

print("\nTotal de homens:", homens)
print("Total de mulheres:", mulheres)
````

````python
total_ate_2000 = 0
total_geral = 0

continuar = "S"

while continuar == "S":

    valor = float(input("Digite o valor do carro: "))
    ano = int(input("Digite o ano do veículo: "))

    if ano <= 2000:
        desconto = valor * 0.12
        total_ate_2000 += 1
    else:
        desconto = valor * 0.07

    valor_final = valor - desconto

    print("Desconto: R$", desconto)
    print("Valor a pagar: R$", valor_final)

    total_geral += 1

    continuar = input("Deseja continuar? (S/N): ").upper()

print("\nTotal de veículos até 2000:", total_ate_2000)
print("Total geral de veículos:", total_geral)
````
