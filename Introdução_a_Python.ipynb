**Como mostrar informações na tela?**

Em Python, utilizamos o comando *print* para mostrar informações na tela, sejam elas texto, números, ou os dois juntos.


**Sintaxe:**

`*print()*`

Deve-se colocar entre parenteses o que quer mostrar na tela, por exemplo o clássico:

`*print('Olá Mundo!') *`

Vamos fazer alguns testes:

```python
print('"Olá Mundo"')
```

**Podemos guardar informações na memória do computador**

Para isso, utilizamos espaços na memória, chamados de Variáveis.

Para guardar um valor dentro de uma variável, utilizamos a seguinte sintaxe:

`variavel = valor`

Algumas coisas importantes!

-> O símbolo **" = "** em python tem significado diferente da matemática. Em python, esse símbolo significa "recebe". Assim, nós lemos o código acima como "variável recebe valor".

-> Onde escrevemos **"variavel"** no codigo, podemos usar qualquer palavra. É o rótulo, ou identificador da nossa variável. É recomendado não usar nomes genéricos.

-> o campo **"valor "** no código, pode ser um número, um texto ou mesmo outra variável.

Vamos ver alguns exemplos:

```python
nome = 'Marcos'
idade = 30

print(f'{nome}: Minha idade é {idade}!!')
```

**E se for necessário que o usuário envie informações?**

Para isso, usamos o comando *input*, cuja sintaxe é parecida com o *print*. Vejam o exemplo:

`nome = input('Insira o Nome:')`

Nesse exemplo, temos o comando input, que mostra a mensagem "*Insira o Nome:*" na tela e aguarda a resposta do usuário.

Antes do comando input, temos "*nome =*", ou seja, o valor que for digitado pelo usuário será armazenado dentro da variável nome.

Vejamos mais exemplos:

```python
nome = input('Insira a Nome: ')
nome
```
