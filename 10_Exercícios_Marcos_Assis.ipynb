```python
v = []

for i in range(16):
    v.append(int(input("Digite um valor: ")))

for i in range(8):
    v[i], v[i + 8] = v[i + 8], v[i]

print(v)
```

```python
v = []

for i in range(20):
    v.append(int(input("Digite um valor: ")))

x = int(input("Digite o valor a procurar: "))

encontrou = False

for i in range(20):
    if v[i] == x:
        print("Encontrado na posição:", i)
        encontrou = True

if not encontrou:
    print("Valor não encontrado")
```

```python
v = []
pares = 0

for i in range(40):
    v.append(int(input("Digite um valor: ")))

for i in range(40):
    if v[i] % 2 == 0:
        pares += 1

print("Quantidade de pares:", pares)
```

```python
v = []

for i in range(40):
    v.append(int(input("Digite um valor: ")))

for i in range(40):
    if v[i] < 0:
        v[i] = 0

print(v)
```

```python
a = []
b = []

for i in range(20):
    a.append(int(input("Vetor A: ")))

for i in range(20):
    b.append(int(input("Vetor B: ")))

c = [0] * 40

ia = 0
ib = 0

for i in range(40):

    if i % 2 == 0:
        c[i] = a[ia]
        ia += 1
    else:
        c[i] = b[ib]
        ib += 1

print(c)
```

```python
v = []

for i in range(40):
    v.append(int(input("Digite um valor: ")))

for i in range(1, 40):
    v[i] = v[i] + v[i - 1]

print(v)
```

```python
frase = input("Digite uma frase: ")

vetor = list(frase)

novo = []

for letra in vetor:
    if letra != " ":
        novo.append(letra)

print("".join(novo))
```

```python
digitos = []

for i in range(5):
    digitos.append(input("Digite um dígito: "))

nomes = {
    "0": "ZERO",
    "1": "UM",
    "2": "DOIS",
    "3": "TRÊS",
    "4": "QUATRO",
    "5": "CINCO",
    "6": "SEIS",
    "7": "SETE",
    "8": "OITO",
    "9": "NOVE"
}

for d in digitos:
    print(nomes[d], end=" ")
```

```python
v1 = []
v2 = []

for i in range(5):
    v1.append(input("Primeira palavra: ")[0])

for i in range(5):
    v2.append(input("Segunda palavra: ")[0])

v1.sort()
v2.sort()

if v1 == v2:
    print("São anagramas")
else:
    print("Não são anagramas")
```
