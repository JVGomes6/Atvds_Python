
Argumentos default(padrão) -> com valores definidos previamente, caso o usuário não insira

```python
def exponencial(base,exp=2):
  resultado=1
  for i in range(exp):
    resultado = resultado*base
  return resultado

print(exponencial(2))   # usando o valor padrão de exp=2, pois não informamos esse valor para a função

print(exponencial(2,10)) #usando o valor de exp=10 que passamos.
```

Argumentos do tipo "lista" são considerados globais, ou seja, alterações dentro de funções vão refletir fora dela

```python
def zeraLista(lista):
  for i in range(len(lista)):
    lista[i] = 0

lista = [10]*10

print(lista)

zeraLista(lista)

print(lista)
```

```python
import math

def volume_esfera(raio):
    volume = (4/3) * math.pi * (raio ** 3)
    return volume

r = float(input("Digite o raio: "))
print(volume_esfera(r))
```

```python
def calcular_media(n1, n2, n3, letra):

    if letra == "A":
        media = (n1 + n2 + n3) / 3

    elif letra == "P":
        media = (n1*5 + n2*3 + n3*2) / 10

    elif letra == "H":
        media = 3 / ((1/n1) + (1/n2) + (1/n3))

    else:
        media = None

    return media

n1 = float(input("Nota 1: "))
n2 = float(input("Nota 2: "))
n3 = float(input("Nota 3: "))
tipo = input("A, P ou H: ").upper()

print(calcular_media(n1, n2, n3, tipo))
```

```python
def primo(n):

    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Digite um número: "))

print(primo(num))
```

```python
import math

def bhaskara(a, b, c):

    delta = b**2 - 4*a*c

    if delta < 0:
        return None, None

    x1 = (-b + math.sqrt(delta)) / (2*a)
    x2 = (-b - math.sqrt(delta)) / (2*a)

    return x1, x2

a = float(input("A: "))
b = float(input("B: "))
c = float(input("C: "))

r1, r2 = bhaskara(a, b, c)

if r1 is None:
    print("Não possui raízes reais")
else:
    print("x1 =", r1)
    print("x2 =", r2)
```

```python
def converter(segundos):

    horas = segundos // 3600
    segundos %= 3600

    minutos = segundos // 60
    segundos %= 60

    return horas, minutos, segundos

tempo = int(input("Digite os segundos: "))

h, m, s = converter(tempo)

print(f"{h}h {m}min {s}s")
```

```python
def idade_dias(anos, meses, dias):

    total = anos * 365
    total += meses * 30
    total += dias

    return total

anos = int(input("Anos: "))
meses = int(input("Meses: "))
dias = int(input("Dias: "))

print("Idade em dias:", idade_dias(anos, meses, dias))
```

```python
def perfeito(n):

    soma = 0

    for i in range(1, n):
        if n % i == 0:
            soma += i

    return soma == n

num = int(input("Digite um número: "))

if perfeito(num):
    print("Número perfeito")
else:
    print("Número não perfeito")
```
